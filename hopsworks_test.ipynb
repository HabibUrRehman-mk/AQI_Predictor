```python
import os
import hopsworks

# 1. Retrieve API Key
api_key = os.getenv("HOPSWORKS_API_KEY")

if not api_key:
    raise ValueError("HOPSWORKS_API_KEY environment variable is not set!")

project_name = "AQI_Predictor_fsd"

# 2. Define a Windows-compatible certificate path
# This creates a 'certs' folder right inside your local project directory
cert_dir = os.path.join(os.getcwd(), "certs")
os.makedirs(cert_dir, exist_ok=True)

# 3. Login specifying the cert_folder
project = hopsworks.login(
    host="eu-west.cloud.hopsworks.ai",
    port=443,
    project=project_name,
    api_key_value=api_key,
    cert_folder=cert_dir  # <--- Fixes the Windows /tmp/ FileNotFoundError
)

print(f"Successfully logged in to project: {project.name}")

# 4. Get the Feature Store
fs = project.get_feature_store()
print("Connected to Feature Store!")
```
